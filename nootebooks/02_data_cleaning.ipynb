```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("C:\\Users\\Ali\\Desktop\\first eda project\\data\\raw__data\\insurance_data_for eda.csv")
```

```python
df.info("index")
```

## Step 1: Removing Unnecessary Columns

The columns 'index' and 'PatientID' do not contribute to analysis since they are only identifiers.  
These columns will be removed.


```python
df = df.drop(columns=["index", "PatientID"])
df.head()
```

## Step 2: Handling Missing Values

We check for missing values in the dataset.  
From the dataset overview:

- age → 5 missing values  
- region → 3 missing values  

These must be handled before analysis.


```python
df.isnull().sum()
```

## Step 3: Imputing Missing Values

- Numerical column (age) → Filled using median.
- Categorical column (region) → Filled using mode.


```python
# Fill age with median
df["age"].fillna(df["age"].median(), inplace=True)

# Fill region with mode
df["region"].fillna(df["region"].mode()[0], inplace=True)

# Verify again
df.isnull().sum()
```

## Step 4: Correcting Data Types

Categorical columns will be converted to category type for better memory efficiency and analysis clarity.


```python
categorical_cols = ["gender", "diabetic", "smoker", "region"]

for col in categorical_cols:
    df[col] = df[col].astype("category")

df.info()
```

## Step 5: Checking for Duplicate Records

Duplicate records may distort analysis and must be removed if present.


```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
# Clean data ko save karein
df.to_csv('final_clean_data.csv', index=False)
print("Data saved. Ab analysis notebook mein ye file load karein.")
```
