## Final Summary of Exploratory Data Analysis
This exploratory data analysis examined the factors influencing insurance claim amounts. The dataset contains demographic and health-related information about individuals along with their medical insurance claims.

Several patterns were identified during the analysis.

Age shows a positive relationship with insurance claims. Older individuals tend to generate higher medical expenses compared to younger individuals.

BMI also plays a significant role in determining healthcare costs. Individuals in higher BMI categories, particularly obese and severely obese groups, show higher claim amounts and a greater probability of extreme medical expenses.

Smoking status has the strongest impact on claim amounts. Smokers consistently show significantly higher claim values than non-smokers, indicating increased health risks associated with smoking.

Chronic health conditions such as diabetes also contribute to increased medical expenses. Individuals diagnosed with diabetes tend to generate higher claims compared to non-diabetic individuals.

Regional differences were also observed in the dataset, with some regions showing slightly higher average claim amounts.

Overall, the analysis highlights that age, BMI, smoking status, and chronic health conditions are the most important factors affecting insurance claim costs.

These insights can help insurance companies improve risk assessment, pricing strategies, and healthcare cost prediction models.
