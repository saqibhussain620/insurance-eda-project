## Multivariate Analysis: Age, BMI and Smoking Status vs Insurance Claim

In this section, we analyze the combined effect of **Age**, **Body Mass Index (BMI)**, and **Smoking Status** on the **insurance claim amount**.

While earlier analyses examined variables individually or in pairs, multivariate analysis allows us to observe how **multiple risk factors interact simultaneously**.

Age reflects the general health risk associated with aging, BMI represents physical health condition, and smoking is a major lifestyle risk factor that significantly affects medical expenses. By visualizing these variables together, we can identify patterns that may influence insurance claim costs.


### Age vs Claim by Smoking Status

This scatter plot visualizes the relationship between **age and insurance claim amount**, while differentiating individuals based on **smoking status**.

Smoking is a critical health risk factor, and this visualization helps determine whether smokers tend to generate higher medical insurance claims compared to non-smokers across different age groups.


```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(9,6))

sns.scatterplot(
    x="age",
    y="claim",
    hue="smoker",
    data=df,
    palette="Set1"
)

plt.title("Age vs Claim by Smoking Status")
plt.xlabel("Age")
plt.ylabel("Claim Amount")

plt.show()
```



**Observation**

The plot clearly shows a strong difference between smokers and non-smokers. Smokers tend to have significantly higher claim amounts compared to non-smokers across almost all age groups.

While claim amounts for non-smokers remain relatively moderate, smokers show a large concentration of high-cost claims. This suggests that smoking is a major driver of insurance costs and increases financial risk for insurance companies.






### BMI vs Claim by Smoking Status

This visualization explores how **Body Mass Index (BMI)** influences insurance claim amounts while also considering smoking behavior.

Higher BMI levels are often associated with obesity and related health conditions. By combining BMI and smoking status, this analysis helps determine whether these two risk factors together lead to higher insurance costs.

```python
plt.figure(figsize=(9,6))

sns.scatterplot(
    x="bmi",
    y="claim",
    hue="smoker",
    data=df,
    palette="coolwarm"
)

plt.title("BMI vs Claim by Smoking Status")
plt.xlabel("BMI")
plt.ylabel("Claim Amount")

plt.show()
```

**Observation**

The scatter plot indicates that individuals with higher BMI values tend to generate higher insurance claims. However, the effect becomes significantly stronger among smokers.

Smokers with high BMI levels are associated with the largest claim amounts in the dataset. In contrast, non-smokers generally show lower claim values even at higher BMI levels.

This suggests that the combination of **high BMI and smoking behavior** creates a much higher health risk, leading to greater medical expenses.


### Combined Impact of Age and BMI on Claim Amount

This bubble plot visualizes three variables simultaneously:
- Age (x-axis)
- BMI (y-axis)
- Claim amount (bubble size)

This visualization helps identify whether individuals with higher BMI and older age tend to generate larger insurance claims.


```python
plt.figure(figsize=(9,6))

sns.scatterplot(
    x="age",
    y="bmi",
    size="claim",
    hue="smoker",
    data=df,
    sizes=(20,400),
    alpha=0.6
)

plt.title("Combined Effect of Age, BMI and Smoking on Claims")

plt.show()
```

**Observation**

The bubble plot highlights the combined influence of age, BMI, and smoking status on insurance claim amounts.

Larger bubbles (representing higher claim amounts) appear more frequently among smokers. Additionally, individuals with both higher BMI and increasing age show larger claim sizes.

This pattern indicates that insurance costs are influenced by the combined effect of multiple risk factors rather than a single variable alone.


### Multivariate Analysis Conclusion

The combined analysis of age, BMI, and smoking status reveals that insurance claim costs are influenced by multiple interacting risk factors.

Smoking emerges as the strongest driver of high claim amounts. Individuals who smoke consistently show significantly larger claims compared to non-smokers.

BMI also contributes to increased claim costs, particularly when combined with smoking. Individuals with higher BMI values tend to incur greater medical expenses.

Age further amplifies these risks, as older individuals with high BMI and smoking habits often produce the largest claim amounts in the dataset.

Overall, the analysis suggests that insurance companies should consider lifestyle factors such as smoking and obesity, along with age, when assessing risk and determining premium pricing.


## Multivariate Analysis: Age, Diabetes Status and Insurance Claim

In this section, we analyze the combined relationship between **Age**, **Diabetes status**, and **Insurance Claim amount**.

Age is an important factor in healthcare costs because medical risks generally increase as people get older. Diabetes is a chronic health condition that may lead to higher medical expenses over time.

By analyzing these variables together, we aim to determine whether diabetic individuals show higher insurance claim amounts across different age groups compared to non-diabetic individuals.


### Age vs Claim by Diabetes Status

This scatter plot shows the relationship between **age and claim amount**, while distinguishing individuals based on whether they are diabetic or not.

This helps identify whether diabetes significantly increases claim costs at different ages.


```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(9,6))

sns.scatterplot(
    x="age",
    y="claim",
    hue="diabetic",
    data=df,
    palette="Set2"
)

plt.title("Age vs Claim by Diabetes Status")
plt.xlabel("Age")
plt.ylabel("Claim Amount")

plt.show()
```

**Observation**

The scatter plot shows that claim amounts generally increase slightly with age. However, the distribution of claim values for diabetic and non-diabetic individuals appears quite similar.

Both groups contain low, medium, and high claim values across different age ranges. This suggests that diabetes alone may not be the primary factor driving large insurance claims in this dataset.


```python
df["age_group"] = pd.cut(
    df["age"],
    bins=[15,25,35,45,55,65],
    labels=["16-25","26-35","36-45","46-55","56-65"]
)
```

### Claim Distribution Across Age Groups by Diabetes Status

This box plot compares claim amounts across different age groups while also considering diabetes status.

The visualization helps determine whether diabetic individuals show higher median claim values within each age group.


```python
plt.figure(figsize=(9,6))

sns.boxplot(
    x="age_group",
    y="claim",
    hue="diabetic",
    data=df
)

plt.title("Claim Amount Across Age Groups by Diabetes Status")
plt.xlabel("Age Group")
plt.ylabel("Claim Amount")

plt.show()
```

**Observation**

The box plot shows that claim amounts increase slightly with age across both diabetic and non-diabetic individuals.

However, the difference between diabetic and non-diabetic groups within each age category is relatively small. This indicates that while age may influence claim costs, diabetes does not appear to drastically change the overall claim distribution in this dataset.


### Claim Distribution by Diabetes Status

This density plot visualizes how claim amounts are distributed for diabetic and non-diabetic individuals.


```python
plt.figure(figsize=(9,6))

sns.kdeplot(
    data=df,
    x="claim",
    hue="diabetic",
    fill=True
)

plt.title("Claim Distribution by Diabetes Status")
plt.xlabel("Claim Amount")

plt.show()
```

**Observation**

The density curves for diabetic and non-diabetic individuals are very similar, indicating that the overall distribution of claim amounts does not differ significantly between the two groups.

Both categories show a concentration of claims around moderate values, with a smaller number of high-cost outliers.


### Heatmap of Average Claim by Age Group and Diabetes Status

This heatmap visualizes the average claim amounts across different combinations of age groups and diabetes status.

Darker colors represent higher average claims, helping identify which group combinations present the highest financial risk.


```python
pivot_table = df.pivot_table(
    values="claim",
    index="age_group",
    columns="diabetic",
    aggfunc="mean"
)

plt.figure(figsize=(8,5))

sns.heatmap(pivot_table, annot=True, cmap="coolwarm")

plt.title("Average Claim Heatmap: Age Group vs Diabetes")
plt.show()
```

### Individual Claim Points by Age and Diabetes Status

This strip plot displays individual claim values for each age while differentiating diabetic and non-diabetic individuals.

It allows us to observe the spread of claim values and identify extreme outliers in the dataset.


```python
plt.figure(figsize=(9,6))

sns.stripplot(
    x="diabetic",
    y="claim",
    hue="age_group",
    data=df,
    jitter=True
)

plt.title("Individual Claim Distribution by Diabetes and Age Group")
plt.xlabel("Diabetes Status")
plt.ylabel("Claim Amount")

plt.show()
```

### Average Claim Amount by Age Group and Diabetes Status

This bar chart compares the average claim amount across different age groups while also considering diabetes status.

It helps identify whether diabetic individuals generate higher average claims within specific age ranges.


```python
plt.figure(figsize=(9,6))

sns.barplot(
    x="age_group",
    y="claim",
    hue="diabetic",
    data=df
)

plt.title("Average Claim by Age Group and Diabetes")
plt.xlabel("Age Group")
plt.ylabel("Average Claim Amount")

plt.show()
```

### Average Claim by Age Group and Diabetes Status

This bar chart illustrates the average insurance claim amounts across different age groups while comparing individuals with and without diabetes.

From the visualization, several patterns emerge:

- In the **16–25 age group**, non-diabetic individuals show higher average claim amounts compared to diabetic individuals.
- In the **26–35 age group**, the claim values for diabetic and non-diabetic individuals are relatively similar.
- In the **36–45 age group**, diabetic individuals show noticeably higher average claim amounts than non-diabetics.
- In the **46–55 age group**, diabetic individuals again have slightly higher claim amounts.
- In the **56–65 age group**, non-diabetic individuals exhibit higher claim amounts compared to diabetics.

Overall, the relationship between diabetes status and claim amount is **not consistent across all age groups**. While diabetes appears to increase claim costs in some middle-age groups, this pattern does not hold for younger and older individuals.

This suggests that **age alone or diabetes alone may not be sufficient predictors of claim amount**, and other factors such as BMI, smoking status, and lifestyle conditions may significantly influence healthcare costs.


### Key Insight

The analysis indicates that claim costs are influenced by a combination of demographic and health factors. While diabetes shows some impact in middle age groups, the effect is not consistent across all ages. This highlights the importance of incorporating additional variables such as BMI, smoking status, and lifestyle indicators when predicting insurance claim amounts.


## Multivariate Analysis: BMI, Gender and Insurance Claim

In this section, we analyze the combined effect of **Body Mass Index (BMI)** and **Gender** on insurance claim amounts.

BMI is an important health indicator that reflects body fat and potential health risks such as obesity-related diseases. Gender differences may also influence healthcare usage and medical expenses.

By analyzing BMI and gender together, we aim to determine whether individuals with higher BMI levels tend to generate larger insurance claims and whether this pattern differs between males and females.


```python
df["bmi_group"] = pd.cut(
    df["bmi"],
    bins=[15,25,30,35,60],
    labels=["Normal","Overweight","Obese","Severely Obese"]
)
```

### Average Claim by BMI Category and Gender

This bar chart compares the average insurance claim amounts across different BMI categories while distinguishing between males and females.

The visualization helps identify whether higher BMI levels lead to increased claim costs and whether this trend differs between genders.


```python
plt.figure(figsize=(9,6))

sns.barplot(
    x="bmi_group",
    y="claim",
    hue="gender",
    data=df
)

plt.title("Average Claim by BMI Category and Gender")
plt.xlabel("BMI Category")
plt.ylabel("Average Claim Amount")

plt.show()
```

### Average Claim by BMI Category and Gender

This bar chart compares the average insurance claim amounts across different BMI categories for males and females.

From the visualization, we can observe a clear relationship between BMI levels and insurance claim costs.

Key observations include:

- Individuals in the **Normal BMI category** have the lowest average claim amounts. Interestingly, females show slightly higher claims than males in this category.
  
- In the **Overweight category**, the average claim amounts for males and females are relatively similar, indicating minimal gender-based variation.

- For the **Obese category**, claim amounts increase significantly for both genders. However, males tend to generate higher average claims compared to females.

- In the **Severely Obese category**, claim costs reach their highest levels. Male individuals in this category exhibit the highest average claim amounts across all BMI groups.

Overall, the visualization suggests that **higher BMI levels are associated with increased insurance claim costs**, regardless of gender. While gender differences exist, the impact of **BMI appears to be a stronger driver of insurance claims**.


### Claim Distribution by BMI Category and Gender

This box plot illustrates the distribution of insurance claims across BMI categories while comparing males and females.

The plot highlights median values, variability, and potential outliers in claim amounts within each BMI category.


```python
plt.figure(figsize=(9,6))

sns.boxplot(
    x="bmi_group",
    y="claim",
    hue="gender",
    data=df
)

plt.title("Claim Distribution by BMI Category and Gender")

plt.show()
```

### Claim Density Across BMI Categories by Gender

The violin plot provides a deeper view of the distribution of claim amounts across BMI categories for both males and females.

It combines box plot information with density estimation to reveal where claim values are most concentrated.


```python
plt.figure(figsize=(9,6))

sns.violinplot(
    x="bmi_group",
    y="claim",
    hue="gender",
    data=df,
    split=True
)

plt.title("Claim Density by BMI Category and Gender")

plt.show()
```

### BMI vs Claim by Gender

This scatter plot visualizes the relationship between BMI and claim amount while distinguishing between male and female individuals.

It helps determine whether claim costs increase as BMI increases and whether this trend differs by gender.


```python
plt.figure(figsize=(9,6))

sns.scatterplot(
    x="bmi",
    y="claim",
    hue="gender",
    data=df
)

plt.title("BMI vs Claim by Gender")

plt.show()
```

### Analysis of Insurance Claims by BMI Category and Gender

To better understand how **BMI and gender influence insurance claim amounts**, we analyzed the data using both a **bar plot** and a **violin plot**. These visualizations provide complementary insights into the average claim values and the distribution of claims across different BMI groups.

---

### 1. Bar Plot Insights (Average Claim Comparison)

The bar plot shows the **average insurance claim amount for males and females across different BMI categories**.

Key observations include:

- Individuals in the **Normal BMI category** have the lowest average claim amounts.
- As BMI increases from **Normal → Overweight → Obese → Severely Obese**, the **average claim amount steadily increases** for both genders.
- In the **Obese and Severely Obese categories**, males tend to have slightly higher average claim amounts compared to females.
- The **Severely Obese group shows the highest average claims**, indicating a strong relationship between high BMI and increased healthcare costs.

Overall, the bar plot highlights a **clear positive relationship between BMI level and average insurance claim cost**.

---

### 2. Violin Plot Insights (Distribution of Claims)

While the bar plot shows average values, the violin plot provides a deeper understanding of the **distribution and density of claims** within each BMI group.

- As BMI categories increase, the violins become **taller and more elongated**, indicating a wider range of claim amounts and a higher likelihood of extreme claims.
- **Females:** Most claim values remain concentrated around **$10,000**, even in higher BMI categories.
- **Males:** In the **Obese and Severely Obese groups**, a secondary concentration of claims appears at much higher values (around **$35,000–$50,000**), suggesting the presence of a **high-cost subgroup**.

---

### 3. Claim Distribution Pattern

The violin plot reveals a **bimodal distribution** in higher BMI categories:

- **Primary cluster:** Most claims fall between **$0 and $15,000**.
- **Secondary cluster:** In the **Obese and Severely Obese groups**, a second cluster of claims appears between **$35,000 and $50,000**.

The inner box plots show that **median claim values remain moderate**, while the long upper tails represent **high-cost outliers reaching $60,000 or more**.

---

### Overall Conclusion

Both visualizations indicate that **BMI is a major factor influencing insurance claim costs**. As BMI increases, not only does the **average claim amount rise**, but the **risk of extremely high claims also becomes more significant**.

While gender differences are visible, particularly in higher BMI categories, the overall pattern suggests that **BMI has a stronger impact on insurance claims than gender alone**.
